A **kernel** is the long-running Python process that executes notebook code cells. VS Code displays

the code.


## 5. Verify the active environment

Python exposes information about the running process through the `sys` module. The most useful
value here is `sys.prefix`, which identifies the active installation or virtual environment.

````python
import platform
import sys
from pathlib import Path

print("Python version:    ", platform.python_version())
print("Python executable: ", sys.executable)
print("Active environment:", sys.prefix)

environment = Path(sys.prefix)
assert environment.name == ".venv", (
    "Select the Rice DSM kernel, restart it, and run this notebook again."
)
````

The active environment should end in this repository's `.venv` directory. The executable may
display uv's underlying managed Python on some systems; `sys.prefix` is the important check for
the environment providing installed packages.

## 6. Run some ordinary Python

A code cell is an interactive Python input. Select the next cell and press `Shift+Enter` or use
the Run button. The kernel evaluates the code and sends its output back to the notebook.

````python
course_name = "Data Science & Machine Learning"
lecture_number = 1

print(f"Welcome to {course_name}, Lecture {lecture_number}!")
````

## 7. The kernel remembers earlier cells

The next cell can use `course_name` because the same kernel process remains alive. If you
restart the kernel and run only the next cell, it will raise `NameError` because the new process
has not executed the cell that defines the name.

````python
words = course_name.split()
print(words)
print("Number of words:", len(words))
````

This persistent state is useful for exploration, but execution order matters. A reproducible
notebook must work in top-to-bottom order from a fresh kernel.

## 8. A reliable notebook habit

Before submitting or sharing a notebook:

1. save your work;
2. restart the kernel;
3. run all cells from the beginning; and
4. read every error rather than assuming old variables still exist.

This habit catches hidden execution-order dependencies.

## 9. If something fails

**`uv` is not recognized:** install `uv`, close and reopen the terminal, then run `uv --version`.

**The project is not found:** run `pwd` and `ls`; navigate to the folder containing
`pyproject.toml`.

**The environment assertion fails:** select **Rice DSM**, restart the kernel, and run all cells.

**A variable is undefined:** run the earlier cell that creates it, or restart and run all.
